```python
import pymupdf

caminho_pdf = "../data/raw/prodist/aren2021956_Prodist_modulo_3_v9.pdf"
doc = pymupdf.open(caminho_pdf)

pagina = doc[14]

texto_bruto = pagina.get_text()
print(texto_bruto)
```

```python
import re

# Remover quebras de linhas que ocorrem no meio de uma frase
def limpar_texto_pdf(texto_bruto):
    texto_limpo = texto_bruto.replace('\n',' ')
    texto_limpo = re.sub(r'\s+', ' ', texto_limpo)
    return texto_limpo.strip()

texto_processado = limpar_texto_pdf(texto_bruto)

print("--- TEXTO LIMPO ---")
print(texto_processado[:500])
```
