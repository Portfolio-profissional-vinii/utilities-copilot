```python
import json
from langchain_text_splitters import RecursiveCharacterTextSplitter

caminho_json = "../data/processed/aren2021956_Prodist_modulo_3_v9_extraido.json"

with open(caminho_json, 'r', encoding='utf-8') as f:
    documento = json.load(f)

texto_teste = documento[14]['texto']
numero_pagina = documento[14]['pagina']

print(f"Tamanho total do texto da página {numero_pagina}: {len(texto_teste)} caracteres.")
```

```python
fatiador = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    length_function=len,
    separators=["\n\n", "\n", "."," ", ""]
)

pedacos = fatiador.split_text(texto_teste)
print(f"A página foi dividida em {len(pedacos)} chunks. \n")
print("-" * 50)

for i, pedaco in enumerate(pedacos[:2]):
    print(f"CHUNK {i + 1} (Tamanho: {len(pedaco)} caracteres)")
    print(pedaco)
    print("-" * 50)
```
