```python
from langchain_huggingface import HuggingFaceEmbeddings

nome_modelo = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"

print("Baixando e carregando o modelo (pode levar alguns minutos na primeira execução...)")
embeddings = HuggingFaceEmbeddings(
    model_name=nome_modelo,
    model_kwargs={'device': 'cuda'},
    encode_kwargs={'normalize_embeddings': True}
)

texto_exemplo = "O medidor apresentou falha no display, gerando faturamento por estimativa."
vetor = embeddings.embed_query(texto_exemplo)

print(f"Sucesso! O texto foi transformado em um vetor de {len(vetor)} dimensões.")
print(f"Aqui estão os primeiros 5 números do seu texto vetorizado:\n{vetor[:5]}")
```
